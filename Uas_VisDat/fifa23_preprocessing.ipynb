# FIFA 23 Data Preprocessing

Notebook preprocessing dataset FIFA 23 untuk analisis dan visualisasi data.

```python
# Import Library

import pandas as pd
import numpy as np
```

```python
# Membaca Dataset

players = pd.read_csv('data/players_fifa23.csv')
teams = pd.read_csv('data/teams_fifa23.csv')
```

## Menampilkan Data Awal

```python
print("===== DATA PLAYERS =====")
print(players.head())

print("\n===== INFO DATASET =====")
print(players.info())

print("\n===== UKURAN DATASET =====")
print(players.shape)
```

## Mengecek Missing Value

```python
print("\n===== MISSING VALUE =====")
print(players.isnull().sum())
```

## Menghapus Kolom Tidak Digunakan

```python
drop_columns = [
    'PhotoUrl'
]

players.drop(
    columns=drop_columns,
    inplace=True,
    errors='ignore'
)
```

## Memilih Kolom Penting

```python
players = players[[
    'Name',
    'Overall',
    'Potential',
    'Age',
    'Club',
    'Nationality',
    'Positions',
    'ValueEUR',
    'WageEUR',
    'PaceTotal',
    'ShootingTotal',
    'PassingTotal',
    'DribblingTotal',
    'DefendingTotal',
    'PhysicalityTotal'
]]
```

## Menghapus Missing Value

```python
players = players.dropna()
```

## Mengecek dan Menghapus Data Duplikat

```python
print("\n===== DATA DUPLIKAT =====")
print(players.duplicated().sum())

players = players.drop_duplicates()
```

## Mengecek Tipe Data

```python
print("\n===== TIPE DATA =====")
print(players.dtypes)
```

## Mengubah Format Data Numerik

```python
numeric_cols = [
    'Overall',
    'Potential',
    'Age',
    'ValueEUR',
    'WageEUR',
    'PaceTotal',
    'ShootingTotal',
    'PassingTotal',
    'DribblingTotal',
    'DefendingTotal',
    'PhysicalityTotal'
]

for col in numeric_cols:

    players[col] = pd.to_numeric(
        players[col],
        errors='coerce'
    )
```

## Menghapus Null Setelah Konversi

```python
players = players.dropna()
```

## Membersihkan Data String

```python
players['Club'] = (
    players['Club']
    .astype(str)
    .str.strip()
)

players['Nationality'] = (
    players['Nationality']
    .astype(str)
    .str.strip()
)

players['Positions'] = (
    players['Positions']
    .astype(str)
    .str.strip()
)
```

## Feature Engineering

```python
# Membuat Kategori Umur

def age_category(age):

    if age < 23:
        return 'Young'

    elif age < 30:
        return 'Adult'

    else:
        return 'Senior'


players['AgeCategory'] = (
    players['Age']
    .apply(age_category)
)

# Membuat Kategori Rating

def rating_category(overall):

    if overall >= 85:
        return 'World Class'

    elif overall >= 75:
        return 'Good Player'

    else:
        return 'Average'


players['RatingCategory'] = (
    players['Overall']
    .apply(rating_category)
)
```

## Filtering Data

```python
players = players[
    players['Overall'] > 65
]
```

## Reset Index

```python
players = players.reset_index(drop=True)
```

## Informasi Dataset Setelah Preprocessing

```python
print("\n===== DATASET SETELAH PREPROCESSING =====")

print(players.head())

print("\n===== UKURAN DATASET BARU =====")
print(players.shape)

print("\n===== INFO DATASET BARU =====")
print(players.info())
```

## Menyimpan Dataset Bersih

```python
players.to_csv(
    'players_fifa23_clean.csv',
    index=False
)

print("\n===================================")
print("PREPROCESSING SELESAI")
print("File berhasil disimpan:")
print("players_fifa23_clean.csv")
print("===================================")
```
